```python
import spacy
import pandas as pd

nlp = spacy.load("en_core_web_sm")

text = """
Satya Nadella is the CEO of Microsoft.
Microsoft is headquartered in Redmond, Washington.
"""

doc = nlp(text)

entities = []

for ent in doc.ents:
    entities.append({
        "Entity": ent.text,
        "Label": ent.label_
    })

df = pd.DataFrame(entities)

print(df)
```

```python
import spacy
from spacy.tokens import Span
from spacy.util import filter_spans

nlp = spacy.load("en_core_web_sm")

text = "John was diagnosed with COVID-19 and Diabetes."

doc = nlp(text)

entities = [ent for ent in doc.ents if ent.text not in ["COVID-19", "Diabetes"]]

for token in doc:
    if token.text in ["COVID-19", "Diabetes"]:
        entities.append(
            Span(doc, token.i, token.i + 1, label="DISEASE")
        )

doc.ents = filter_spans(entities)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
